```python
import numpy as np
import matplotlib.pyplot as plt
import nibabel as nib

print("MRI analysis notebook is ready!")
```

```python
# Load MRI file
img = nib.load("../data/brain_mri.nii.gz")

# Convert MRI into NumPy array
mri_data = img.get_fdata()

# Display dimensions
print("MRI dimensions:", mri_data.shape)
```

```python
# Extract first volume from 4D MRI
volume = mri_data[..., 0]

print("3D MRI dimensions:", volume.shape)
```

```python
# Select middle slice
slice_index = volume.shape[2] // 2

# Extract 2D slice
mri_slice = volume[:, :, slice_index]

# Display MRI
plt.figure(figsize=(6, 6))
plt.imshow(mri_slice.T, cmap="gray", origin="lower")
plt.title("Brain MRI - Axial Slice")
plt.axis("off")
plt.show()
```

```python
fig, axes = plt.subplots(2, 3, figsize=(12, 8))

slice_indices = np.linspace(
    0, volume.shape[2] - 1, 6, dtype=int
)

for ax, idx in zip(axes.flat, slice_indices):

    ax.imshow(
        volume[:, :, idx].T,
        cmap="gray",
        origin="lower"
    )

    ax.set_title(f"Slice {idx}")
    ax.axis("off")

plt.suptitle("Brain MRI - Multiple Axial Slices")
plt.tight_layout()
plt.show()
```
